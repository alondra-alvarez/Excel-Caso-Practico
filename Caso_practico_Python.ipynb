#Investigacion previa sobre el tema

**Que es la tuberculosis**  
La tuberculosis (TB) es una enfermedad infecciosa causada por una bacteria llamada Mycobacterium tuberculosis. Afecta principalmente a los pulmones, aunque también puede atacar otras partes del cuerpo como los riñones, la columna vertebral o el cerebro.

**Causas de la tuberculisis**  
La causa principal es la infección por la bacteria Mycobacterium tuberculosis, que se transmite de persona a persona por el aire, generalmente cuando una persona con TB pulmonar activa:

* Tose
* Estornuda
* Habla
* Escupe

Al inhalar estas gotitas contaminadas, otra persona puede infectarse.
**Factores de riesgo**  

Aunque cualquiera puede infectarse, hay ciertos factores que aumentan el riesgo:

* Tener un sistema inmunológico debilitado (por VIH, diabetes, tratamientos con esteroides o quimioterapia)
* Desnutrición
* Vivir o trabajar en lugares cerrados y con poca ventilación
* Ser fumador
* Consumo excesivo de alcohol o drogas
* Vivir en zonas con alta prevalencia de TB

**Tipos de tuberculosis**  
1. **TB latente**: La bacteria está en el cuerpo pero no causa síntomas y no es contagiosa.
2. **TB activa**: La bacteria se multiplica y causa síntomas y puede transmitirse a otros.

**¿Por qué ocurre esto?**
Varias posibles causas:

* Los hombres tienden a buscar atención médica más tarde, lo que retrasa el diagnóstico
* Factores de riesgo más comunes en hombres, como tabaquismo y consumo de alcohol, aumentan su vulnerabilidad .
* Ocupaciones con exposición elevada (como minería o trabajos en espacios cerrados) también hacen suponer mayor exposición en hombres .
* También hay discusión sobre diferencias biológicas que podrían influir en la susceptibilidad .

**metodos de diagnostico**

| Abreviatura | Significado                    | Explicación sencilla                                                                                   |
| ----------- | ------------------------------ | ----------------------------------------------------------------------------------------------------- |
| **rel**     | Recaída                        | La persona tuvo tuberculosis antes y la enfermedad volvió a aparecer. Se detecta con exámenes médicos.|
| **sn**      | Baciloscopia pulmonar negativa | No se ven bacterias en la muestra de pulmón al microscopio, pero puede haber tuberculosis; se hacen más pruebas para estar seguros. |
| **sp**      | Baciloscopia pulmonar positiva | Se ven bacterias en la muestra de pulmón al microscopio; es un caso típico y se empieza tratamiento.  |
| **ep**      | Extrapulmonar                  | La tuberculosis está en otras partes del cuerpo fuera del pulmón; el diagnóstico es más complicado y usa varios exámenes. |
            

**tipos de tratamiento**
**1. Tuberculosis sensible a fármacos (TB DS)**
- **Duración:** 6 meses
- **Fases:**
  - Fase intensiva (2 meses): Rifampicina, Isoniazida, Pirazinamida, Etambutol (RHZE)
  - Fase de continuación (4 meses): Rifampicina e Isoniazida (RH)
- Tratamiento estándar recomendado por la OMS.

**2. Tuberculosis multirresistente (MDR-TB)**
- Resistencia a rifampicina e isoniazida.
- **Duración:** 6 a 20 meses o más.
- Uso de medicamentos de segunda línea como bedaquilina, linezolid, levofloxacino, delamanid, entre otros.
- Requiere monitoreo riguroso.

**3. Tuberculosis extensivamente resistente (XDR-TB)**
- Resistente a rifampicina, isoniazida y a los medicamentos de segunda línea más importantes.
- Tratamientos más largos y complejos.
- Alta tasa de fracaso terapéutico y mortalidad.

**Índices de muertes por tuberculosis (estimaciones globales)**

| Año   | Muertes estimadas (millones) | Observaciones                         |
|-------|------------------------------|-------------------------------------|
| 2021  | ≈1.4 millones                | TB es una de las principales causas infecciosas de muerte. |
| 2023  | Datos preliminares indican leve descenso, pero sigue siendo alta la mortalidad en regiones con TB-MDR y coinfección con VIH. |

- La mayoría de las muertes ocurren en Asia del Sur, África Subsahariana y en personas con coinfección VIH.
- La mortalidad es mayor en casos de TB resistente debido a tratamiento más difícil.

# Modificar la estructura de los datos

```python
who.head()
```

```python
population.head()
```

```python
# Convertir las columnas despues de iso3 en filas
who2 = who.melt(id_vars=["country", "year", "iso2", "iso3"])
who2.head(10)
```

```python
# Añadir la columna que indica el genero
who2["gender"] = np.where(who2["variable"].str.contains("m"),"masculino","femenino")
who2.head(100)
```

```python
# funcion para asignar el grupo de edad
def asignar_groupedad(valor):
  if "014" in valor:
     return "0-14"
  elif "1524" in valor:
    return "15-24"
  elif "25-34" in valor:
    return "25-34"
  elif "3544" in valor:
    return "35-44"
  elif "4554" in valor:
    return "45-54"
  elif "5564" in valor:
    return "55-64"
  elif "65" in valor:
    return "65+"
```

```python
who2["agegroup"] = who2["variable"].apply(asignar_groupedad)
who2.head()
```

```python
# funcion para asignar los metodos de diagnóstico
def asignar_metodo(valor):
  if "rel" in valor:
    return "recaida"
  if "sn" in valor:
    return "esputo pulmonar negativo"
  if "sp" in valor:
    return "esputo pulmonar positivo"
  if "ep" in valor:
    return "extrapulmonar"
```

```python
#asignar la funcion a una columna
who2["method"] = who2["variable"].apply(asignar_metodo)
who2.head()
```

```python
#eliminar la columna variable que ya no me sirve
who2 = who2.drop(columns=["variable"])
```

```python
who2.head()
```

```python
who2 = who2.rename(columns={"value":"cases"})
```

```python
who2.head()
```

```python
# vamos a realizar un join con population
df = pd.merge(left=who2, right=population, how="inner", on=["country", "year"])
df.head(100)
```

```python
df.to_csv("tuberculosis.csv", index=False)
```
