## Tipo de datos y format

````python
# Enteros (int)
edad = 26
print(type(edad))

# Decimales (float)
decimal = 3.1416
print(type(decimal))

# Cadenas de texto (str)
nombre = "Eduardo"
print(type(nombre))

#Booleanos (Bool)
verdadero = True
print(type(verdadero))
````

````python
# Listas:: Colecction ordenada de datos
numeros = [1,2,3]
print(type(numeros))

# Tuplas:: coleccion ordenada de datos,NO SE PUEDE MODIFICAR
tupla = (4,5,6)
print(type(tupla))

#Diccionarios: colleciones de pares con estructura llave-valor
persona = {"nombre":"Alondra","apellido":"Alvarez",
           "frutas_favoritas":["manzana","uva"]}
print(type(persona))

#Salidas y formato
nombre = "Alondra"
print("nombre:",nombre)

print("Hola {} cómo estas?".format(nombre))


nombre = "Alondra"
respuesta = "Super_Bien"
print("nombre:",nombre)

print("Hola {} cómo estas? {}".format(nombre, respuesta))

# F-strings
print(f"Hola {nombre} como estas? {respuesta}")

pi = 3.1416
print(f"Pi es igual a {pi:.2f}")
````

## Controles de flujo

**IF,ELSE** Permite ejecutar codigo solo si se se cumple una condición

````python
edad = 100
if edad >= 18:
  print("eres mayor de edad")
else:
  print("no eres mayor de edad")
````

**WHILE** Ejecuta un bloque de codigo mientras se cumpla una condicion

````python
contador = 0
while contador < 6:
  print(contador)
  contador = contador +1
````

**FOR** Permite repetir un bloque de código un número determinado de veces o sobre una colección

````python
for i in range(1,7):
  print(i)
````

````python
# Los FOR rambien pueden recorrer listas
frutas_favoritas = ["mango","manzana","uva","papaya"]
for fruta in frutas_favoritas:
  print(fruta)
````

**TRY,EXCEPT** Permite atrapar errores para que el programa no se detenga si courre uno

````python
numero = "abc" # type error
try:
 multiplicacion = numero ** 2
 print(multiplicacion)

except:
  print("El tipo de dato no es numerico")
````

1. Usa if else para decir si un numero es impar o no

Modulo de un numero:  
10/5, modulo = 0
11/5, modulo = 1



````python
#Modulo de un numero
11 % 5

#Problema: identificar si un numero es impar o no
# Pseudocodigo
# 1. Definir un numero
# 2. Si el modulo de 2 es 0,entonces es un numero par
# 3. de lo contrario, es un numero impar

numero = 28
if numero % 2 == 0:
  print("el numero es par")
else:
  print ("El numero es impar")

# input() - permite que el usuario introduzca un valor
numero = int(input("Introduce un numero"))

if numero % 2 == 0:
  print("El numero es par")
else:
  print("El numero es impar")
````

## Ejercicio 3

3. Crea un while que cuente cuantas veces un usuario escribo algo antes de escribir "salir"

````python
# pseodocodigo
# 1. Definir un contador,que incremente de uno en 1
# 2. Definir una palabra
# 3. Iniciar el while mientras la palabra sea distinta de "salir"
# 4. Si la palabra no es salir,aumentamos uno mas en el contador
# 5. pedir al usuario que introduzca una nueva palabra
# 6. Imprimir el contador

contador = 0
palabra = input("introduce una palabra ")

while palabra != "salir":
  contador = contador +1
  palabra = input ("introduce otra palabra ")

print(f"el contador final es de {contador}")
````
